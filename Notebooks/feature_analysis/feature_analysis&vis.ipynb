```python
def replace_featureticklabels(orig_ticks):
    replace_dict = {
        'bond_breaking_cnt': 'Bond Breaking \n Number',
        'bond_formation_cnt': 'Bond Formation \n Number', 
        'reaction_barrier': 'Reaction Barrier',
        'reaction_center_cnt': 'Reaction Center \n Number',
        'is_react_on_rings_binary': 'Is React \n on Rings',
        'ts_volume': 'TS Volume',
        'rotatable_bond_cnt': 'Rotatable Bond \n Number',
        'ts_fragment_cnt': 'TS Fragment \n Number'
    }
    replaced_ticks = [replace_dict[str] for str in orig_ticks]
    return replaced_ticks

def replace_errorticklabels(orig_ticks):
    replace_dict = {
        'bond_breaking_error': 'Error Number of \n Bond Breaking',
        'bond_formation_error': 'Error Number of \n Bond Formation',
        'reaction_center_error': 'Error Number of \n Reaction Center',
        'mean_angle_error': 'Mean Angle Error',
        'max_angle_error': 'Max Angle Error',
        'mean_torsional_error': 'Mean Torsional Error',
        'max_torsional_error': 'Max Torsional Error',
        'mean_improper_error': 'Mean Improper Error',
        'max_improper_error': 'Max Improper Error'
    }
    replaced_ticks = [replace_dict[str] for str in orig_ticks]
    return replaced_ticks
```

```python
def preprocess_separate_data(feature_df, error_df, binary_is_react=True):
    feature_processed = feature_df.copy()
    
    for feature in feature_types.keys():
        if feature in feature_processed.columns:
            feature_processed = feature_processed[feature_processed[feature] != -1]
    
    if 'is_react_on_rings' in feature_processed.columns:
        if binary_is_react:
            feature_processed['is_react_on_rings_binary'] = feature_processed['is_react_on_rings'].apply(
                lambda x: 1 if x > 0 else 0 
            )
            
            feature_processed = feature_processed.drop(columns=['is_react_on_rings'])
    
    error_processed = error_df.copy()
    for error in error_types.keys():
        if error in error_processed.columns:
            if -1 in error_processed[error].values:
                error_processed = error_processed[error_processed[error] != -1]
    
    common_idx = feature_processed.index.intersection(error_processed.index)
    feature_aligned = feature_processed.loc[common_idx]
    error_aligned = error_processed.loc[common_idx]
    
    return feature_aligned, error_aligned
```

```python
def calculate_feature_error_correlation(feature_df, error_df):
    feature_vars = []
    for feature in feature_types.keys():
        if feature == 'is_react_on_rings':
            if 'is_react_on_rings_binary' in feature_df.columns:
                feature_vars.append('is_react_on_rings_binary')
        elif feature in feature_df.columns:
            feature_vars.append(feature)
    
    error_vars = list(error_types.keys())
    
    corr_matrix = pd.DataFrame(index=feature_vars, columns=error_vars)
    pvalue_matrix = pd.DataFrame(index=feature_vars, columns=error_vars)
    
    for feature in feature_vars:
        for error in error_vars:
            if feature in feature_df.columns and error in error_df.columns:
                feature_data = feature_df[feature].dropna()
                error_data = error_df[error].dropna()
                
                common_idx = feature_data.index.intersection(error_data.index)
                if len(common_idx) < 3: 
                    corr = np.nan
                    pval = np.nan
                else:
                    feature_aligned = feature_data.loc[common_idx]
                    error_aligned = error_data.loc[common_idx]
                    
                    if feature == 'is_react_on_rings_binary':
                        corr, pval = stats.pointbiserialr(feature_aligned, error_aligned)
                    
                    elif ('continuous' in error_types.get(error, '')):
                        if feature in ['reaction_barrier', 'ts_volume']:
                            corr, pval = stats.pearsonr(feature_aligned, error_aligned)
                        else:
                            corr, pval = stats.spearmanr(feature_aligned, error_aligned)
                    
                    else:
                        corr, pval = stats.spearmanr(feature_aligned, error_aligned)
                
                corr_matrix.loc[feature, error] = corr
                pvalue_matrix.loc[feature, error] = pval
    
    corr_matrix = corr_matrix.astype(float)
    pvalue_matrix = pvalue_matrix.astype(float)
    
    return corr_matrix, pvalue_matrix, feature_vars, error_vars
```

```python
def plot_feature_error_correlation_comparison(feature1_df, error1_df, feature2_df, error2_df, 
                                             method1_name="method 1", method2_name="method 2", threshold=0.1):

    feature1_aligned, error1_aligned = preprocess_separate_data(
        feature1_df, error1_df, binary_is_react=True
    )
    feature2_aligned, error2_aligned = preprocess_separate_data(
        feature2_df, error2_df, binary_is_react=True
    )
    
    corr1, pval1, feature_vars1, error_vars1 = calculate_feature_error_correlation(
        feature1_aligned, error1_aligned
    )
    corr2, pval2, feature_vars2, error_vars2 = calculate_feature_error_correlation(
        feature2_aligned, error2_aligned
    )
    
    all_features = sorted(set(feature_vars1).union(set(feature_vars2)))
    all_errors = sorted(set(error_vars1).union(set(error_vars2)))
    
    corr1 = corr1.reindex(index=all_features, columns=all_errors)
    pval1 = pval1.reindex(index=all_features, columns=all_errors)
    corr2 = corr2.reindex(index=all_features, columns=all_errors)
    pval2 = pval2.reindex(index=all_features, columns=all_errors)

    ticklabel_features = replace_featureticklabels(corr1.index)
    ticklabel_errors = replace_errorticklabels(corr1.columns)

    width = 180 / 25.4
    height = 80 / 25.4
    figs = plt.figure(figsize=(width, height))
    
    axes = figs.subplots(1, 2)
    
    vmin = min(corr1.min().min(), corr2.min().min())
    vmax = max(corr1.max().max(), corr2.max().max())
    vcenter = 0
    vabs = max(abs(vmin), abs(vmax))
    
    def create_annotation_matrix(corr_matrix, pval_matrix):
        annot_matrix = []
        for i in range(len(corr_matrix.index)):
            row = []
            for j in range(len(corr_matrix.columns)):
                r = corr_matrix.iloc[i, j]
                p = pval_matrix.iloc[i, j]
                
                if np.isnan(r) or np.isnan(p):
                    row.append("")
                else:
                    if p < 0.001:
                        p_str = "$p$<0.001"
                    elif p < 0.01:
                        p_str = f"$p$={p:.3f}"
                    elif p < 0.05:
                        p_str = f"$p$={p:.3f}"
                    else:
                        p_str = f"$p$={p:.3f}"
                    
                    row.append(f"$r$={r:.2f}\n{p_str}")
            annot_matrix.append(row)
        return np.array(annot_matrix)
    
    annot1 = create_annotation_matrix(corr1, pval1)
    annot2 = create_annotation_matrix(corr2, pval2)
    
    heatmap1 = sns.heatmap(corr1, 
                          ax=axes[0],
                          cmap='RdBu_r',
                          center=0,
                          vmin=-vabs,
                          vmax=vabs,
                          square=False,
                          cbar_kws={"shrink": 0.8},
                          annot=annot1,  
                          fmt="", 
                          annot_kws={'size':4, "ha": "center", "va": "center"},
                          linewidths=0.5,
                          linecolor='gray',
                          xticklabels=ticklabel_errors,
                          yticklabels=ticklabel_features)
    
    cbar = heatmap1.collections[0].colorbar

    cbar.ax.set_title('$r$', pad=8, y=1.02)
    
    axes[0].set_title(f'{method1_name}: Features vs Errors', fontweight='bold')
    axes[0].set_xlabel('Errors')
    axes[0].set_ylabel('Features')

    plt.setp(axes[0].get_xticklabels(), 
         rotation=60, 
         ha="right",
         rotation_mode="anchor")

    
    heatmap2 = sns.heatmap(corr2, 
                          ax=axes[1],
                          cmap='RdBu_r',
                          center=0,
                          vmin=-vabs,
                          vmax=vabs,
                          square=False,
                          cbar_kws={"shrink": 0.8},
                          annot=annot2,
                          fmt="",
                          annot_kws={'size':4, "ha": "center", "va": "center"},
                          linewidths=0.5,
                          linecolor='gray',
                          xticklabels=ticklabel_errors,
                          yticklabels=ticklabel_features)
    
    cbar = heatmap2.collections[0].colorbar

    cbar.ax.set_title('$r$', pad=8, y=1.02)
    
    axes[1].set_title(f'{method2_name}: Features vs Errors', fontweight='bold')
    axes[1].set_xlabel('Errors')
    axes[1].set_ylabel('Features')
    plt.setp(axes[1].get_xticklabels(), 
         rotation=60, 
         ha="right",
         rotation_mode="anchor")

    n_rows, n_cols = corr2.shape
    
    plt.tight_layout()
    
    return {
        'corr1': corr1,
        'pval1': pval1,
        'corr2': corr2,
        'pval2': pval2,
        'feature1': feature1_aligned,
        'error1': error1_aligned,
        'feature2': feature2_aligned,
        'error2': error2_aligned,
        'feature1_orig': feature1_df,  
        'feature2_orig': feature2_df   
    }
```

```python
def identify_strong_feature_error_correlations(corr1, pval1, corr2, pval2, threshold=0.1):
    strong_pairs = []
    
    features = corr1.index.tolist()
    errors = corr1.columns.tolist()
    
    for feature in features:
        vis_feature = 'is_react_on_rings' if feature == 'is_react_on_rings_binary' else feature
        
        for error in errors:
            corr1_val = abs(corr1.loc[feature, error])
            corr2_val = abs(corr2.loc[feature, error])
            pval1_val = pval1.loc[feature, error]
            pval2_val = pval2.loc[feature, error]
            
            if np.isnan(corr1_val) or np.isnan(corr2_val):
                continue
            
            if (corr1_val >= threshold or corr2_val >= threshold and 
                pval1_val < 0.05 and pval2_val < 0.05):
                
                direction1 = "positive" if corr1.loc[feature, error] > 0 else "negative"
                direction2 = "positive" if corr2.loc[feature, error] > 0 else "negative"
                
                strong_pairs.append({
                    'feature': feature,
                    'vis_feature': vis_feature,
                    'error': error,
                    'corr1': corr1.loc[feature, error],
                    'corr2': corr2.loc[feature, error],
                    'pval1': pval1_val,
                    'pval2': pval2_val,
                    'direction1': direction1,
                    'direction2': direction2,
                    'avg_corr': (abs(corr1_val) + abs(corr2_val)) / 2
                })
    
    strong_pairs.sort(key=lambda x: x['avg_corr'], reverse=True)
    
    return strong_pairs
```

```python
replace_dict_features = {
    'bond_breaking_cnt': 'Bond Breaking Number',
    'bond_formation_cnt': 'Bond Formation Number', 
    'reaction_barrier': 'Reaction Barrier',
    'reaction_center_cnt': 'Reaction Center Number',
    'is_react_on_rings_binary': 'Is React on Rings',
    'is_react_on_rings': 'Is React on Rings',
    'ts_volume': 'TS Volume',
    'rotatable_bond_cnt': 'Rotatable Bond Number',
    'ts_fragment_cnt': 'TS Fragment Number'
}

replace_dict_errors = {
        'bond_breaking_error': 'Error Number of Bond Breaking',
        'bond_formation_error': 'Error Number of Bond Formation',
        'reaction_center_error': 'Error Number of Reaction Center',
        'mean_angle_error': 'Mean Angle Error',
        'max_angle_error': 'Max Angle Error',
        'mean_torsional_error': 'Mean Torsional Error',
        'max_torsional_error': 'Max Torsional Error',
        'mean_improper_error': 'Mean Improper Error',
        'max_improper_error': 'Max Improper Error'
}

replace_dict_features2 = {
    'bond_breaking_cnt': 'Bond Breaking Number',
    'bond_formation_cnt': 'Bond Formation Number', 
    'reaction_barrier': 'Reaction Barrier (eV)',
    'reaction_center_cnt': 'Reaction Center Number',
    'is_react_on_rings_binary': 'Is React on Rings',
    'is_react_on_rings': 'Is React on Rings',
    'ts_volume': 'TS Volume (Å³)',
    'rotatable_bond_cnt': 'Rotatable Bond Number',
    'ts_fragment_cnt': 'TS Fragment Number'
}

replace_dict_errors2 = {
        'bond_breaking_error': 'Error Number of Bond Breaking',
        'bond_formation_error': 'Error Number of Bond Formation',
        'reaction_center_error': 'Error Number of Reaction Center',
        'mean_angle_error': 'Mean Angle Error (Degree)',
        'max_angle_error': 'Max Angle Error (Degree)',
        'mean_torsional_error': 'Mean Torsional Error (Degree)',
        'max_torsional_error': 'Max Torsional Error (Degree)',
        'mean_improper_error': 'Mean Improper Error (Degree)',
        'max_improper_error': 'Max Improper Error (Degree)'
}

replace_is_on_ring_labels = {
    0: 'Not On Rings',
    1: 'On Aromatic Rings',
    2: 'On Non-aromatic Rings'
}

def select_plot_type(feature_type, error_type):

    feature_type_map = {
        'bond_breaking_cnt': 'discrete',
        'bond_formation_cnt': 'discrete',
        'reaction_barrier': 'continuous',
        'reaction_center_cnt': 'discrete',
        'is_react_on_rings': 'categorical',  
        'ts_volume': 'continuous',
        'rotatable_bond_cnt': 'discrete',
        'ts_fragment_cnt': 'discrete'
    }
    
    error_type_map = {
        'bond_breaking_error': 'discrete',
        'bond_formation_error': 'discrete',
        'reaction_center_error': 'discrete',
        'mean_angle_error': 'continuous',
        'max_angle_error': 'continuous',
        'mean_torsional_error': 'continuous',
        'max_torsional_error': 'continuous',
        'mean_improper_error': 'continuous',
        'max_improper_error': 'continuous'
    }
    
    feature_type = feature_type_map.get(feature_type, feature_type)
    error_type = error_type_map.get(error_type, error_type)

    if feature_type == 'categorical' and error_type == 'continuous':
        return 'violin' 
    elif feature_type == 'categorical' and error_type == 'discrete':
        return 'stacked_bar'  
    elif feature_type == 'discrete' and error_type == 'continuous':
        return 'box'  
    elif feature_type == 'discrete' and error_type == 'discrete':
        return 'heatmap' 
    elif feature_type == 'continuous' and error_type == 'continuous':
        return 'scatter'  
    elif feature_type == 'continuous' and error_type == 'discrete':
        return 'scatter_discrete'  
    else:
        return 'scatter' 
    
def plot_strong_feature_error_correlations(strong_pairs, feature1_df, error1_df, feature2_df, error2_df,
                                          feature1_orig=None, feature2_orig=None,
                                          method1_name="method 1", method2_name="method 2", top_n=6):

    top_pairs = strong_pairs[:top_n]
    
    if not top_pairs:
        return None
    
    if feature1_orig is not None:
        feature1_vis = feature1_orig
        feature2_vis = feature2_orig
    else:
        feature1_vis = feature1_df
        feature2_vis = feature2_df
    
    n_plots = min(len(top_pairs), top_n)
    n_cols = 3
    n_rows = (n_plots + n_cols - 1) // n_cols

    width = 180 / 25.4
    height_single = 50 / 25.4
    
    fig = plt.figure(figsize=(width, height_single * n_rows))
    axes = fig.subplots(n_rows, n_cols)
    axes = axes.flatten() if n_plots > 1 else [axes]
    
    colors = ['#FF6B6B', '#4ECDC4']  
    colors_light = ['#FFB6B6', '#B2EBE2']  
    
    for idx, pair in enumerate(top_pairs):
        if idx >= len(axes):
            break
            
        ax = axes[idx]
        feature = pair['vis_feature'] 
        error = pair['error']
        
        if feature == 'is_react_on_rings' and feature in feature1_vis.columns:
            data1_x = feature1_vis[feature]
            feature_type = 'categorical'
        elif feature in feature1_df.columns:
            data1_x = feature1_df[feature]
            feature_type = 'continuous' if feature in ['reaction_barrier', 'ts_volume'] else 'discrete'
        else:
            continue
            
        data1_y = error1_df[error] if error in error1_df.columns else None
        
        if feature == 'is_react_on_rings' and feature in feature2_vis.columns:
            data2_x = feature2_vis[feature]
            feature_type = 'categorical'
        elif feature in feature2_df.columns:
            data2_x = feature2_df[feature]
        else:
            continue
            
        data2_y = error2_df[error] if error in error2_df.columns else None
        
        if data1_x is None or data1_y is None or data2_x is None or data2_y is None:
            continue
        
        common_idx1 = data1_x.dropna().index.intersection(data1_y.dropna().index)
        common_idx2 = data2_x.dropna().index.intersection(data2_y.dropna().index)
        
        error_type = 'continuous' if 'error' in error and any(x in error for x in ['mean', 'max', 'angle', 'torsional', 'improper', 'volume']) else 'discrete'
        
        plot_type = select_plot_type(feature, error)
        print(f"feature: {feature}({feature_type}), error: {error}({error_type}) -> chart type: {plot_type}")
        
        if plot_type == 'violin':
            plot_violin_comparison(ax, feature1_vis, error1_df, feature2_vis, error2_df, 
                                  feature, error, method1_name, method2_name, colors)
            
        elif plot_type == 'stacked_bar':
            plot_stacked_bar_comparison(ax, feature1_vis, error1_df, feature2_vis, error2_df,
                                       feature, error, method1_name, method2_name, colors, colors_light)
            
        elif plot_type == 'box':
            plot_box_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                               feature, error, method1_name, method2_name, colors, feature_type)
            
        elif plot_type == 'heatmap':
            plot_scatter_comparison_discrete_discrete(ax, feature1_df, error1_df, feature2_df, error2_df,
                                   feature, error, method1_name, method2_name, colors)
            # plot_heatmap_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                                #    feature, error, method1_name, method2_name)
        elif plot_type == 'scatter_discrete':
            plot_scatter_comparison_continuous_discrete(ax, feature1_df, error1_df, feature2_df, error2_df,
                                   feature, error, method1_name, method2_name, colors, colors_light)
            
        elif plot_type == 'scatter':
            plot_scatter_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                                   feature, error, method1_name, method2_name, colors, colors_light)
        
        ax.legend(loc='best')
        title = f'{replace_dict_features[pair["feature"]]} vs {replace_dict_errors[error]}\n'
        # title += f'{method1_name}: r={pair["corr1"]:.3f} '
        # title += f'p={pair["pval1"]:.3f}\n'
        # title += f'{method2_name}: r={pair["corr2"]:.3f} '
        # title += f'p={pair["pval2"]:.3f}'
        
        ax.set_title(title)
        
        ax.grid(True, alpha=0.3, zorder=0)
    
    for idx in range(len(top_pairs), len(axes)):
        axes[idx].set_visible(False)
    
    plt.tight_layout()
    return fig

def plot_violin_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                          feature, error, method1_name, method2_name, colors):

    data_list = []
    
    all_cats = sorted(set(feature1_df[feature].unique()) | set(feature2_df[feature].unique()))
    
    for cat in all_cats:
        if cat != -1: 
            cat_data1 = error1_df.loc[feature1_df[feature] == cat, error]
            for val in cat_data1.values:
                data_list.append({
                    'category': replace_is_on_ring_labels[cat] if feature == 'is_react_on_rings' else f'{cat}',
                    'value': val,
                    'method': method1_name 
                })
            
            cat_data2 = error2_df.loc[feature2_df[feature] == cat, error]
            for val in cat_data2.values:
                data_list.append({
                    'category': replace_is_on_ring_labels[cat] if feature == 'is_react_on_rings' else f'{cat}',
                    'value': val,
                    'method': method2_name
                })
    
    df = pd.DataFrame(data_list)
    
    sns.violinplot(data=df, x='category', y='value', hue='method',
                   ax=ax, palette=colors, split=True,
                   inner='quartile', cut=0, scale='width')
    
    ax.legend(loc='best')
    ax.set_xticklabels(ax.get_xticklabels(), rotation=10, ha='center')
    
    ax.set_xlabel(f'{replace_dict_features2[feature]}')
    ax.set_ylabel(replace_dict_errors2[error])
    
    return ax

def plot_stacked_bar_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                               feature, error, method1_name, method2_name, colors, colors_light):

    categories = sorted([c for c in set(feature1_df[feature].unique()) | set(feature2_df[feature].unique()) if c != -1])
    error_values = sorted([e for e in set(error1_df[error].unique()) | set(error2_df[error].unique()) if not np.isnan(e)])
    

    data1 = []
    data2 = []
    
    for cat in categories:

        cat_data1 = error1_df.loc[feature1_df[feature] == cat, error]
        counts1 = []
        for err_val in error_values:
            counts1.append((cat_data1 == err_val).sum())
        data1.append(counts1)
        

        cat_data2 = error2_df.loc[feature2_df[feature] == cat, error]
        counts2 = []
        for err_val in error_values:
            counts2.append((cat_data2 == err_val).sum())
        data2.append(counts2)
    

    data1 = np.array(data1)
    data2 = np.array(data2)
    

    x = np.arange(len(categories))
    width = 0.35
    

    bottom1 = np.zeros(len(categories))
    for i in range(len(error_values)):
        ax.bar(x - width/2, data1[:, i], width, bottom=bottom1, 
               color=colors_light[0], edgecolor=colors[0], label=f'{method1_name}: {error}={error_values[i]}' if i == 0 else "")
        bottom1 += data1[:, i]
    

    bottom2 = np.zeros(len(categories))
    for i in range(len(error_values)):
        ax.bar(x + width/2, data2[:, i], width, bottom=bottom2,
               color=colors_light[1], edgecolor=colors[1], label=f'{method2_name}: {error}={error_values[i]}' if i == 0 else "")
        bottom2 += data2[:, i]
    
    ax.set_xticks(x)
    ax.set_xticklabels([f'{c}' for c in categories])
    ax.set_xlabel(f'{replace_dict_features2[feature]}')
    ax.set_ylabel(f'{replace_dict_errors2[error]}')
    

    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=colors_light[0], edgecolor=colors[0], label=method1_name),
        Patch(facecolor=colors_light[1], edgecolor=colors[1], label=method2_name)
    ]
    ax.legend(handles=legend_elements, loc='best')

def plot_box_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                       feature, error, method1_name, method2_name, colors, feature_type):
    data_list1 = []
    data_list2 = []
    categories = []
    
    all_vals = sorted([v for v in set(feature1_df[feature].unique()) | set(feature2_df[feature].unique()) if v != -1 and not np.isnan(v)])
    
    for val in all_vals:
        val_data1 = error1_df.loc[feature1_df[feature] == val, error]
        if len(val_data1) > 0:
            data_list1.append(val_data1.values)
            categories.append(f'{val}')
        
        val_data2 = error2_df.loc[feature2_df[feature] == val, error]
        if len(val_data2) > 0:
            data_list2.append(val_data2.values)
    
    positions1 = np.arange(len(data_list1)) * 3 - 0.5
    positions2 = np.arange(len(data_list2)) * 3 + 0.5
    
    bp1 = ax.boxplot(data_list1, positions=positions1, label=method1_name,
                   widths=0.6, patch_artist=True,
                   boxprops=dict(facecolor=colors[0], alpha=0.6),
                   medianprops=dict(color='darkred', linewidth=2),
                   whiskerprops=dict(color='darkred'),
                   capprops=dict(color='darkred'),
                #    flierprops=dict(marker='o', color='darkred', alpha=0.5),
                   showfliers=False)
    
    bp2 = ax.boxplot(data_list2, positions=positions2, label=method2_name,
                   widths=0.6, patch_artist=True,
                   boxprops=dict(facecolor=colors[1], alpha=0.6),
                   medianprops=dict(color='darkcyan', linewidth=2),
                   whiskerprops=dict(color='darkcyan'),
                   capprops=dict(color='darkcyan'),
                #    flierprops=dict(marker='^', color='darkcyan', alpha=0.5),
                   showfliers=False)
    
    ax.set_xticks(np.arange(len(categories)) * 3)
    ax.set_xticklabels(categories)
    ax.set_xlabel(f'{replace_dict_features2[feature]} ')
    ax.set_ylabel(replace_dict_errors2[error])

def plot_scatter_comparison_discrete_discrete(ax, feature1_df, error1_df, feature2_df, error2_df,
                           feature, error, method1_name, method2_name, colors):
    
    feature_vals = sorted([v for v in set(feature1_df[feature].unique()) | set(feature2_df[feature].unique()) 
                          if v != -1 and not np.isnan(v)])
    error_vals = sorted([e for e in set(error1_df[error].unique()) | set(error2_df[error].unique()) 
                        if not np.isnan(e)])
    
    feature_val_to_idx = {val: i for i, val in enumerate(feature_vals)}
    error_val_to_idx = {val: i for i, val in enumerate(error_vals)}
    
    scatter_data1_x = []
    scatter_data1_y = []
    scatter_data2_x = []
    scatter_data2_y = []
    
    for idx in range(len(feature1_df)):
        f_val = feature1_df[feature].iloc[idx]
        e_val = error1_df[error].iloc[idx]
        
        if f_val != -1 and not np.isnan(f_val) and not np.isnan(e_val):
            y_pos = error_val_to_idx[e_val] + np.random.uniform(-0.2, 0.2)  
            x_pos = feature_val_to_idx[f_val] + np.random.uniform(-0.2, -0.05) 
            
            scatter_data1_x.append(x_pos)
            scatter_data1_y.append(y_pos)
    
    for idx in range(len(feature2_df)):
        f_val = feature2_df[feature].iloc[idx]
        e_val = error2_df[error].iloc[idx]
        
        if f_val != -1 and not np.isnan(f_val) and not np.isnan(e_val):
            y_pos = error_val_to_idx[e_val] + np.random.uniform(-0.2, 0.2) 
            x_pos = feature_val_to_idx[f_val] + np.random.uniform(0.05, 0.2)
            
            scatter_data2_x.append(x_pos)
            scatter_data2_y.append(y_pos)
    
    scatter1 = ax.scatter(scatter_data1_x, scatter_data1_y, 
                         alpha=0.7, s=20, label=method1_name,
                         color=colors[0], marker='o', edgecolors='white', linewidth=0.5)
    
    scatter2 = ax.scatter(scatter_data2_x, scatter_data2_y, 
                         alpha=0.7, s=20, label=method2_name,
                         color=colors[1], marker='^', edgecolors='white', linewidth=0.5)
    
    ax.set_ylabel(replace_dict_errors2[error])
    ax.set_xlabel(replace_dict_features2[feature])
    
    ax.set_yticks(range(len(error_vals)))
    ax.set_yticklabels(error_vals)

    ax.set_xticks(range(len(feature_vals)))
    ax.set_xticklabels(feature_vals)
    
    ax.grid(True, alpha=0.3, linestyle='--')
    
    ax.set_title(f'{replace_dict_features[feature]} vs {replace_dict_errors[error]}')
    
    ax.set_xlim(-0.5, len(feature_vals) - 0.5)
    ax.set_ylim(-0.5, len(error_vals) - 0.5)

def plot_scatter_comparison_continuous_discrete(ax, feature1_df, error1_df, feature2_df, error2_df,
                           feature, error, method1_name, method2_name, colors, colors_light):
    
    common_idx1 = feature1_df[feature].dropna().index.intersection(error1_df[error].dropna().index)
    common_idx2 = feature2_df[feature].dropna().index.intersection(error2_df[error].dropna().index)
    
    x1 = feature1_df.loc[common_idx1, feature]
    y1_raw = error1_df.loc[common_idx1, error]
    x2 = feature2_df.loc[common_idx2, feature]
    y2_raw = error2_df.loc[common_idx2, error]
    
    jitter_strength = 0.15
    
    y1_jittered = y1_raw + np.random.uniform(-jitter_strength, -0.05, size=len(y1_raw))
    y2_jittered = y2_raw + np.random.uniform(0.05, jitter_strength, size=len(y2_raw))
    
    ax.scatter(x1, y1_jittered,
              alpha=0.7, color=colors[0], label=method1_name,
              edgecolors=colors[0], s=10)
    
    ax.scatter(x2, y2_jittered,
              alpha=0.7, color=colors[1], label=method2_name,
              edgecolors=colors[1], s=10, marker='^')
    
    ax.set_xlabel(f'{replace_dict_features2[feature]} ')
    ax.set_ylabel(replace_dict_errors2[error])

def plot_scatter_comparison(ax, feature1_df, error1_df, feature2_df, error2_df,
                           feature, error, method1_name, method2_name, colors, colors_light):

    common_idx1 = feature1_df[feature].dropna().index.intersection(error1_df[error].dropna().index)
    common_idx2 = feature2_df[feature].dropna().index.intersection(error2_df[error].dropna().index)
    

    ax.scatter(feature1_df.loc[common_idx1, feature], error1_df.loc[common_idx1, error],
              alpha=0.7, color=colors_light[0], label=method1_name,
              edgecolors=colors_light[0], s=10)
    

    ax.scatter(feature2_df.loc[common_idx2, feature], error2_df.loc[common_idx2, error],
              alpha=0.7, color=colors_light[1], label=method2_name,
              edgecolors=colors_light[1], s=10, marker='^')
    
    
    ax.set_xlabel(f'{replace_dict_features2[feature]} ')
    ax.set_ylabel(replace_dict_errors2[error])
```

```python
def analyze_methods_feature_error_comparison(feature1_df, error1_df, feature2_df, error2_df,
                                            method1_name="method 1", method2_name="method 2", threshold=0.1, top_n=9):

    result = plot_feature_error_correlation_comparison(
        feature1_df, error1_df, feature2_df, error2_df, 
        method1_name, method2_name, threshold=threshold
    )
    
    plt.savefig('feature_error_correlation_matrices.pdf', dpi=1200, bbox_inches='tight')
    plt.show()
    

    strong_pairs = identify_strong_feature_error_correlations(
        result['corr1'], result['pval1'], 
        result['corr2'], result['pval2'],
        threshold=threshold
    )
    
    if strong_pairs:
        fig = plot_strong_feature_error_correlations(
            strong_pairs,
            result['feature1'], result['error1'],
            result['feature2'], result['error2'],
            feature1_orig=feature1_df,  
            feature2_orig=feature2_df,
            method1_name=method1_name, 
            method2_name=method2_name,
            top_n=top_n
        )
        
        if fig is not None:
            plt.savefig('strong_feature_error_correlations.pdf', dpi=1200, bbox_inches='tight')
            plt.show()

    
    return {
        'strong_pairs': strong_pairs,
        'corr_matrices': {
            method1_name: result['corr1'],
            method2_name: result['corr2']
        },
        'pvalue_matrices': {
            method1_name: result['pval1'],
            method2_name: result['pval2']
        },
        'processed_data': {
            'feature1': result['feature1'],
            'error1': result['error1'],
            'feature2': result['feature2'],
            'error2': result['error2']
        }
    }
```
