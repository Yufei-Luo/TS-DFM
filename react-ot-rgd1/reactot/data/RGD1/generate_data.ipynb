```python
import numpy as np
import pickle
```

```python
import h5py
import argparse
```

```python
def generate_data(datapath, datasplit):

    assert datasplit in [
        "id_train",
        "id_val",
        "id_test",
        "ood_test_type",
        "ood_test_size"
    ]
    with open(datapath, 'rb') as f:
        datalist = pickle.load(f)
    
    data = datalist[datasplit]

    res = {'reactant': {'positions': [], 'charges': [], 'num_atoms': [], 'ediff': []},
            'product': {'positions': [], 'charges': [], 'num_atoms': [], 'ediff': []},
            'transition_state': {'positions': [], 'charges': [], 'num_atoms': []}}
    for i in range(len(data)):
        res['reactant']['positions'].append(data[i]['R_G'])
        res['product']['positions'].append(data[i]['P_G'])
        res['transition_state']['positions'].append(data[i]['TS_G'])

        res['reactant']['charges'].append(data[i]['atom_types'])
        res['product']['charges'].append(data[i]['atom_types'])
        res['transition_state']['charges'].append(data[i]['atom_types'])

        res['reactant']['num_atoms'].append(len(data[i]['atom_types']))
        res['product']['num_atoms'].append(len(data[i]['atom_types']))
        res['transition_state']['num_atoms'].append(len(data[i]['atom_types']))

        res['reactant']['ediff'].append(data[i]['TS_E'])
        res['product']['ediff'].append(data[i]['TS_E'])

    return res
```

```python
train_data = generate_data('/home/yufeiluo/research/dataset/RGD1/RGD1_subset.pickle', 'id_train')
valid_data = generate_data('/home/yufeiluo/research/dataset/RGD1/RGD1_subset.pickle', 'id_val')
test_data = generate_data('/home/yufeiluo/research/dataset/RGD1/RGD1_subset.pickle', 'id_test')
test_data_ood_size = generate_data('/home/yufeiluo/research/dataset/RGD1/RGD1_subset.pickle', 'ood_test_size')
test_data_ood_type = generate_data('/home/yufeiluo/research/dataset/RGD1/RGD1_subset.pickle', 'ood_test_type')
with open('train.pkl', 'wb') as f:
    pickle.dump(train_data, f)
with open('valid.pkl', 'wb') as f:
    pickle.dump(valid_data, f)
with open('test.pkl', 'wb') as f:
    pickle.dump(test_data, f)
with open('test_ood_size.pkl', 'wb') as f:
    pickle.dump(test_data_ood_size, f)
with open('test_ood_type.pkl', 'wb') as f:
    pickle.dump(test_data_ood_type, f)
```
